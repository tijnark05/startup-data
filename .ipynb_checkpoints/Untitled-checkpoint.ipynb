```python
# Cell 1: Import all required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# Set visualization style
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette("husl")

print("Libraries imported successfully!")
print(f"Pandas version: {pd.__version__}")
print(f"NumPy version: {np.__version__}")
```

# Data Preprocessing

```python
# Cell 6: Data Cleaning and Preprocessing
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split

print("="*60)
print("DATA PREPROCESSING")
print("="*60)

# Create a copy
df_clean = df.copy()

# 1. Encode categorical variables
categorical_cols = ['investor_type', 'sector', 'founder_background']
le_dict = {}

for col in categorical_cols:
    le = LabelEncoder()
    df_clean[f'{col}_encoded'] = le.fit_transform(df_clean[col])
    le_dict[col] = le
    print(f"Encoded {col}: {len(le.classes_)} unique values")

# 2. Encode target
target_encoder = LabelEncoder()
df_clean['outcome_encoded'] = target_encoder.fit_transform(df_clean['outcome'])
print(f"\nTarget classes: {target_encoder.classes_}")

# 3. Select features
feature_cols = ['funding_rounds', 'founder_experience_years', 'team_size', 
                'market_size_billion', 'product_traction_users', 'burn_rate_million',
                'revenue_million', 'investor_type_encoded', 'sector_encoded', 
                'founder_background_encoded']

X = df_clean[feature_cols]
y = df_clean['outcome_encoded']

print(f"\nFeatures shape: {X.shape}")
print(f"Target shape: {y.shape}")

# 4. Handle outliers using IQR
def handle_outliers(df, col):
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    df[col] = df[col].clip(lower_bound, upper_bound)
    return df

# Apply to numerical columns
for col in ['funding_rounds', 'team_size', 'burn_rate_million', 'revenue_million']:
    X = handle_outliers(X, col)

print("\nOutliers handled!")

# 5. Scale features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled = pd.DataFrame(X_scaled, columns=feature_cols)

print(f"\nScaled features shape: {X_scaled.shape}")

# 6. Split data
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42, stratify=y
)

print(f"\nTrain size: {X_train.shape[0]}")
print(f"Test size: {X_test.shape[0]}")
print(f"Train class distribution: {np.bincount(y_train)}")
print(f"Test class distribution: {np.bincount(y_test)}")
```

#  Train Multiple Models

```python
# Cell 7: Train and Compare Multiple Models
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report, confusion_matrix

# Initialize models
models = {
    'Logistic Regression': LogisticRegression(max_iter=1000, random_state=42),
    'Random Forest': RandomForestClassifier(n_estimators=100, random_state=42),
    'XGBoost': XGBClassifier(n_estimators=100, random_state=42, eval_metric='mlogloss'),
    'SVM': SVC(kernel='rbf', probability=True, random_state=42)
}

# Train and evaluate
results = {}
predictions = {}

print("="*60)
print("MODEL TRAINING AND EVALUATION")
print("="*60)

for name, model in models.items():
    print(f"\nTraining {name}...")
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    
    # Store predictions
    predictions[name] = y_pred
    
    # Calculate metrics
    results[name] = {
        'Accuracy': accuracy_score(y_test, y_pred),
        'Precision': precision_score(y_test, y_pred, average='weighted'),
        'Recall': recall_score(y_test, y_pred, average='weighted'),
        'F1-Score': f1_score(y_test, y_pred, average='weighted')
    }
    
    print(f"{name} - Accuracy: {results[name]['Accuracy']:.4f}")
    print(f"{name} - F1-Score: {results[name]['F1-Score']:.4f}")

# Compare models
print("\n" + "="*60)
print("MODEL COMPARISON")
print("="*60)
results_df = pd.DataFrame(results).T
display(results_df.round(4))

# Plot comparison
fig, ax = plt.subplots(figsize=(12, 6))
results_df.plot(kind='bar', ax=ax, rot=0)
ax.set_title('Model Performance Comparison', fontsize=14)
ax.set_ylabel('Score')
ax.set_xlabel('Model')
ax.legend(loc='lower right')
plt.tight_layout()
plt.show()
```

# Best Model - XGBoost Detailed Analysis

```python
# Cell 8: Best Model (XGBoost) Detailed Analysis
from sklearn.metrics import confusion_matrix, roc_curve, roc_auc_score
from sklearn.preprocessing import label_binarize

# Get the best model (XGBoost)
best_model = models['XGBoost']
y_pred = predictions['XGBoost']

print("="*60)
print("XGBOOST MODEL - DETAILED EVALUATION")
print("="*60)

# Classification report
print("\nClassification Report:")
print(classification_report(y_test, y_pred, target_names=target_encoder.classes_))

# Confusion Matrix
cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=target_encoder.classes_,
            yticklabels=target_encoder.classes_)
plt.title('Confusion Matrix - XGBoost', fontsize=14)
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()

# Feature Importance
if hasattr(best_model, 'feature_importances_'):
    importance_df = pd.DataFrame({
        'feature': feature_cols,
        'importance': best_model.feature_importances_
    }).sort_values('importance', ascending=False)
    
    plt.figure(figsize=(10, 8))
    plt.barh(importance_df['feature'], importance_df['importance'])
    plt.xlabel('Importance')
    plt.title('XGBoost Feature Importance', fontsize=14)
    plt.tight_layout()
    plt.show()
    
    print("\nTop 10 Most Important Features:")
    display(importance_df.head(10))
```

# Cross-Validation

```python
# Cell 9: Cross-Validation
from sklearn.model_selection import cross_val_score, StratifiedKFold

print("="*60)
print("CROSS-VALIDATION")
print("="*60)

# Perform 5-fold cross-validation
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
cv_scores = cross_val_score(best_model, X_scaled, y, cv=cv, scoring='f1_weighted')

print(f"Cross-validation F1 scores: {cv_scores}")
print(f"Mean CV F1 score: {cv_scores.mean():.4f}")
print(f"Std CV F1 score: {cv_scores.std():.4f}")

# Plot CV results
plt.figure(figsize=(8, 5))
plt.bar(range(1, 6), cv_scores, color='skyblue', edgecolor='black')
plt.axhline(y=cv_scores.mean(), color='red', linestyle='--', label=f'Mean: {cv_scores.mean():.4f}')
plt.xlabel('Fold')
plt.ylabel('F1 Score')
plt.title('Cross-Validation F1 Scores', fontsize=14)
plt.legend()
plt.ylim(0, 1)
plt.show()
```

# Hyperparameter Tuning

```python
# Cell 10: Hyperparameter Tuning with GridSearchCV
from sklearn.model_selection import GridSearchCV

print("="*60)
print("HYPERPARAMETER TUNING")
print("="*60)

# Define parameter grid
param_grid = {
    'n_estimators': [100, 200],
    'max_depth': [3, 5, 7],
    'learning_rate': [0.01, 0.1, 0.3],
    'subsample': [0.8, 1.0]
}

# Create XGBoost model
xgb = XGBClassifier(random_state=42, eval_metric='mlogloss')

# Grid search
grid_search = GridSearchCV(
    xgb, 
    param_grid, 
    cv=3, 
    scoring='f1_weighted',
    n_jobs=-1,
    verbose=1
)

print("Starting grid search... (this may take a few minutes)")
grid_search.fit(X_train, y_train)

print(f"\nBest parameters: {grid_search.best_params_}")
print(f"Best CV F1 score: {grid_search.best_score_:.4f}")

# Train final model
best_xgb = grid_search.best_estimator_
best_xgb.fit(X_train, y_train)

# Evaluate on test set
y_pred_tuned = best_xgb.predict(X_test)

print("\nTuned Model Performance:")
print(classification_report(y_test, y_pred_tuned, target_names=target_encoder.classes_))

# Compare tuned vs untuned
print("\nPerformance Improvement:")
print(f"Before Tuning F1: {results['XGBoost']['F1-Score']:.4f}")
print(f"After Tuning F1: {f1_score(y_test, y_pred_tuned, average='weighted'):.4f}")
```

# ROC Curves

```python
# Cell 11: ROC Curves
from sklearn.preprocessing import label_binarize

# Binarize the output
y_test_bin = label_binarize(y_test, classes=[0, 1, 2])
n_classes = 3

# Get predicted probabilities
y_pred_proba = best_xgb.predict_proba(X_test)

# Plot ROC curves
plt.figure(figsize=(10, 8))

for i in range(n_classes):
    fpr, tpr, _ = roc_curve(y_test_bin[:, i], y_pred_proba[:, i])
    roc_auc = roc_auc_score(y_test_bin[:, i], y_pred_proba[:, i])
    plt.plot(fpr, tpr, label=f'{target_encoder.classes_[i]} (AUC = {roc_auc:.4f})')

plt.plot([0, 1], [0, 1], 'k--', label='Random Classifier')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curves - XGBoost', fontsize=14)
plt.legend(loc='lower right')
plt.grid(True)
plt.show()
```

# Save the model

```python
# Cell 12: Save the trained model
import joblib
import pickle

print("="*60)
print("SAVING MODEL")
print("="*60)

# Save model
joblib.dump(best_xgb, 'startup_failure_model.pkl')
print("✅ Model saved as 'startup_failure_model.pkl'")

# Save scaler and encoders
joblib.dump(scaler, 'scaler.pkl')
joblib.dump(target_encoder, 'target_encoder.pkl')
joblib.dump(le_dict, 'label_encoders.pkl')
print("✅ Scaler and encoders saved")

# Create a prediction function
def predict_startup(features_df):
    """
    Predict startup outcome
    
    Args:
        features_df: DataFrame with raw features
    
    Returns:
        prediction: Predicted class (IPO, Acquired, Failed)
        probabilities: Probability for each class
    """
    # Scale features
    features_scaled = scaler.transform(features_df[feature_cols])
    
    # Predict
    pred = best_xgb.predict(features_scaled)
    probs = best_xgb.predict_proba(features_scaled)
    
    return target_encoder.inverse_transform(pred)[0], probs[0]

print("\n✅ Prediction function created!")
print("Example usage:")
print("prediction, probabilities = predict_startup(sample_data)")
```

# Test Predictions

```python
# Cell 13: Test the prediction function with sample data
print("="*60)
print("TESTING PREDICTION FUNCTION")
print("="*60)

# Create a sample startup
sample = pd.DataFrame({
    'funding_rounds': [3],
    'founder_experience_years': [15],
    'team_size': [50],
    'market_size_billion': [100],
    'product_traction_users': [500000],
    'burn_rate_million': [5],
    'revenue_million': [1000],
    'investor_type_encoded': [0],  # tier1_vc
    'sector_encoded': [0],          # AI
    'founder_background_encoded': [1]  # ex_bigtech
})

prediction, probabilities = predict_startup(sample)

print(f"Predicted Outcome: {prediction}")
print("\nProbabilities:")
for i, cls in enumerate(target_encoder.classes_):
    print(f"  {cls}: {probabilities[i]*100:.2f}%")

# Test with random samples from test set
print("\n" + "="*60)
print("PREDICTIONS ON TEST SAMPLES")
print("="*60)

test_samples = X_test.iloc[:5]
preds = best_xgb.predict(test_samples)
actual = y_test[:5]

for i in range(5):
    print(f"Sample {i+1}:")
    print(f"  Predicted: {target_encoder.inverse_transform([preds[i]])[0]}")
    print(f"  Actual: {target_encoder.inverse_transform([actual.iloc[i]])[0]}")
    print()
```
